#### 보너스 문제 1
혜화동-대학동-잠실3동의 연령별 1인가구수와 외출-커뮤니케이션이 모두 적은 집단의 관계를 다른 방식으로 자유롭게 시각화하고 해석해주세요.
시각화 방식: 연령대, 지역의 combination에 대해 back to back barplot

- 혜화동에서는 20대 1인가구가 가장 많이 분포되어 있으나 커뮤니케이션, 외출 분포를 보면 70대 인구가 높게 나타난다. 즉, 혜화동 내에서는 소외된 노인이 1인가구에 살면서 외부 활동을 하지 못하고 있다는 문제를 제기할 수 있다.
- 대학동에서는 시간이 흐를수록 1인가구의 연령대별 분포는 유사하지만, 젊은 연령대의 커뮤니케이션 적은 인구 수가 많아지고 있다. 학기가 시작된 이후 은둔청년(은둔생활을 하는 청년)이  많아지는 것으로 해석할 수 있다.
- 잠실3동에서는 혜화동과 반대의 양상을 보여준다. 1인가구수 비율은 70대가 가장 많지만 커뮤니케이션이 적은 인구는 2~40대에서 약간 더 높은 수치를 기록한다. 즉, 노년층은 1인가구이더라도 커뮤니케이션, 외출 등의 활동을 하고 있으나, 2~40대의 경우 100명 중 절반 이상이 외부 활동을 하지 않는 것으로 파악된다. 잠실3동에서 제시할 수 있는 문제 상황은 고령화보다는 청년층, 중년층의 외부 활동이 너무 적다는 것이다.

````python
from matplotlib import gridspec

def barh_plot_by_dong(df, ax1, ax2, loc_nm, col_nms):
    df_loc_m = df[(df['행정동명'] == loc_nm) & (df['성별'] == 1)]
    df_loc_f = df[(df['행정동명'] == loc_nm) & (df['성별'] == 2)]

    # Plot for col_nms[0] on the first axis
    ax1.barh(df_loc_m['연령대'], -df_loc_m[col_nms[0]], color='blue', label=f'남자 ({col_nms[0]})')
    ax1.barh(df_loc_f['연령대'], -df_loc_f[col_nms[0]], color='pink', left = -df_loc_m[col_nms[0]], label=f'여자 ({col_nms[0]})')
    ax1.set_title(f'{loc_nm} - {col_nms[0]}', fontsize=10)
    
    # Plot for col_nms[1] on the second axis
    ax2.barh(df_loc_m['연령대'], df_loc_m[col_nms[1]], color='cyan', label=f'남자 ({col_nms[1]})')
    ax2.barh(df_loc_f['연령대'], df_loc_f[col_nms[1]], color='red', left = df_loc_m[col_nms[1]], label=f'여자 ({col_nms[1]})')
    ax2.set_title(f'{loc_nm} - {col_nms[1]}', fontsize=10)

    # Show '연령대' only on the first plot
    ax1.set_ylabel('연령대')
    ax2.set_yticklabels([])  # Hide y-axis labels on the second subplot
````

````python
dong_list = ["혜화동", "대학동", "잠실3동"]
col_nms = ["1인가구수", "외출-커뮤니케이션이 모두 적은 집단(전체)"]

# Create figure and gridspec with larger figsize
fig = plt.figure(figsize=(20, 30))
fig.suptitle(f"지역별 {col_nms[0]}와 {col_nms[1]}", fontsize = 20,
            position = (0.5, 1.0-0.05))
    
outer_grid = gridspec.GridSpec(len(df_list), len(dong_list), wspace=0.2, hspace=0.3)

# Add a dictionary to store all legends (labels and colors)
handles, labels = [], []

for i, df in enumerate(df_list):
    for j, dong in enumerate(dong_list):
        # Create a GridSpec inside each outer cell for the two subplots
        inner_grid = gridspec.GridSpecFromSubplotSpec(1, 2, subplot_spec=outer_grid[i, j], wspace=0.1)

        # Create the two subplots within this grid
        # One for 1인가구수, and the other for 외출-커뮤니케이션
        ax1 = fig.add_subplot(inner_grid[0])
        ax2 = fig.add_subplot(inner_grid[1])
        barh_plot_by_dong(df, ax1, ax2, loc_nm=dong, col_nms=col_nms)

        # Collect handles and labels from the first plot
        h1, l1 = ax1.get_legend_handles_labels()
        h2, l2 = ax2.get_legend_handles_labels()

        # Only add unique labels
        handles.extend(h1 + h2)
        labels.extend(l1 + l2)

# Create a single legend for the figure
fig.legend(handles[:4], labels[:4], loc='upper center', ncol=1, bbox_to_anchor=(0.9, 0.93))
plt.tight_layout() 
plt.show()
````

> XGBoost는 트리 모델의 일종으로 Gradient Boosting과 같이 Gradient를 이용해 이전 모형을 개선하는 방향으로 학습이 진행되며 과적합 방지를 위한 파라미터가 추가된다. <br>
초기값 설정 이후 Gradient와 Hessian을 얻고 Regression Tree에 적합시킨다. Gradient Boosting과 다르게 병렬 학습이 가능하여 GPU를 활용할 수 있다.


Boost Parameters:
- n_estimators[default=100] : 트리 모델의 수
- early_stopping_rounds : valid loss에 개선이 없다면 Early Stop. n_estimators가 높게 설정하고 early_stopping_rounds를 설정하여 과적합을 방지하는 전략으로 사용된다.
- learning_rate[default=0.1] : Gradient Descent에서 사용되는 학습률
- min_child_weight[default=1] : child 에서 필요한 모든 관측치에 대한 가중치의 최소합이 값보다 샘플 수가 작으면 leaf node가 된다. 너무 크게 설정하면 Under-fitting 될 수 있다.
- max_depth[default=6] : 트리의 최대 깊이
- gamma[default=0] : 트리에서 분기를 추가적으로 할지 말지 결정하는 최소 손실 감소 값. 값을 크게 설정할 수록 과적합이 감소하는 효과가 있다.
- subsample[default=1] : 각 트리마다 데이터 샘플링 비율 over-fitting방지. 일반적으로 0.5 ~ 1을 사용
- colsample_bytree[default=1] : XGBoost는 Column Sampling을 진행하여 트리가 생성되는데 이때 사용되는 파라미터로 각 트리마다 feature 샘플링 비율을 지칭한다. 일반적으로 0.5 ~ 1을 사용한다.
- reg_lambda [default=1] : L2 규제(Ridge)
- reg_alpha [default=0] : L1 규제(Lasso)
- scale_pos_weight[default = 1] : 클래스 불균형일 때 사용. 일반적으로 (# of neg / # of pos) 값으로 설정한다. 

Train Task Parameters:
- objective: 손실함수 지정(lienar, logistic, softmax, softprob, ...)
- eval_metric: 평가 지표(rmse, mae, logloss, auc, ...)

#### 문제.9
Cross Validation에 대해 간단하게 설명해주세요

<p>CV는 데이터를 여러 부분으로 나누고, 각 부분을 Train과 Validation 용도로 번갈아 사용하여 모델을 평가하는 방법이다.

Train 데이터가 많지 않을 때 사용할 수 있으며 교차 검증을 통해 모델의 일반화 성능을 더 정확하게 측정할 수 있고 하이퍼 파라미터 튜닝 또한 가능하다. <br>

일반적으로 K-Fold CV가 사용되는데 K개의 폴드 세트를 만들어서 K번의 Train과 Validation을 끝낸 후, 각 K개의 Validation 결과를 평균 낸 것이 최종 Validation 결과이다. <br>

일반적인 교차 검증은 데이터를 랜덤하게 섞지만, 이 경우 훈련 데이터의 클래스 분포와 테스트 데이터의 클래스 분포가 크게 차이날 수 있다. <br>

특히, 클래스 불균형이 있는 경우 이런 문제가 크게 발생할 수 있는데 이를 위해 사용되는 기법이 Stratified K Fold 이다. <br>

이는 train/validation 분할 시 클래스 분포를 따라 분할하여 학습 데이터와 검증 데이터의 클래스 분포가 유사하도록 fold를 생성한다. 
</p>

## Bonus 2
XGB의 Feature importance는 트리기반 알고리즘을 사용하는 모델에 대해 적용되는 방법이다. 트리 모델은 불순도(impurity)를 최소화 하는 방향으로 분기를 한다. 트리 분기 시 분기할 node를 선택하는데 이 때 information gain을 이용된다. Information gain은 (상위 노드의 불순도 - 분기한 후 좌/우 노드의 불순도) 를 계산한 것으로, information gain이 크면 노드가 분기했을 때 불순도가 많이 감소한다는 것을 의미한다. 즉, 분기 후 불순도를 많이 낮추는 방향으로 분기가 진행되며 이는 information 측면에서 더 많은 정보를 얻었다고 해석할 수 있다.  

Feature importance는 해당 feature가 불순도를 얼마나 감소시켰는지를 기준으로 계산된다. 즉, 트리기반 모델에서는 feature가 불순도를 많이 감소시킬수록 importance가 커진다. 하지만, Feature Importance는 노드가 분기 할때의 정보 이득만을 고려하여 중요도를 부여하기 때문에 과적합에 대해 고려하지 못한다. 

또한, Feature Importance는 모델 학습의 결과이기 때문에 단순히 Feature Importance가 높은 변수만을 이용할 때 동일한 성능을 기대할 수 없다. 분기 과정에서 Featrue Importance가 높은 변수들은 초기 분기에 빈번하게 쓰이고, Feature Importance가 낮은 변수들이 Leaf Node 근처에서 자주 쓰이는 경우 Feature Importance가 낮은 변수들이 오히려 최종적인 분기과정에서 주요한 역할을 하기 때문에 해당 변수들을 제외시키고 모델을 학습시키는 것은 모델의 성능을 저하시킬 수 있다. 

Feature Importance가 높아서 해당 변수들이 모델의 성능을 높인다고 결론 짓는 것은 인과관계를 잘못 해석한 것이라고 판단할 수 있다.

Feature Importance는 Feature Selection과 다르기 때문에 변수 선택을 위해서는 Best Subset Selection, Forward Selection, Backward Selection 등을 선택하는 것이 바람직하다
